Just like the binary represention whenever we want to display the type of any octal number it will display the 'str'

The oct() function returns a string because it is designed to format a number into a human-readable representation using the 0o prefix, which requires text characters. If it returned an integer, Python would simply display it in the default decimal (base 10) format, losing the octal visual representation you requested.

```python
y = oct(x)
print(y)
print(type(y))
```

```python
print("The radix of the octal represent are: 0o and 0O.")
My_octal_number = 0o323476432
print(My_octal_number)
```

```python
num = 3242
octal = oct(num)
print(octal)
print(f"{num:o}") # Returns '6252' instead of '0o6252'
```
