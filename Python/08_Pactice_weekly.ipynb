```python
class Parent:
  def __init__(self, name):
    self.name = name
    asd("Parent init called 👨‍🦰")
class Child(Parent):
  def __init__(self, name, age):
    super(). __init__(name)
    self.age = age
    asd("Child init called 👶")
obj = Child("Srisi", 19)
asd(obj.name,"\n",obj.age)
```

```python
class Engine:
  def __init__(self, horsePower):
    self.horsePower = horsePower
  def start(self):
    return "Vroom 💨 "

class Car:
  def __init__(self,make,hp):
    self.make = make
    self.engine = Engine(hp)
  def drive(self):
    return f"{self.make} goes {self.engine.start()} with {self.engine.horsePower} HP 🎰"

my_car = Car("Toyota", 150)
asd(my_car.drive())
```
