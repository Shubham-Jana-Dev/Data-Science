```python
class My_name:
  def __init__ (self,name):
    self.name = name
  def __len__(self):
    i = 0
    for j in self.name:
      i += 1
    return i

name_obj = My_name("Shubham")
print(len(name_obj.name))
```

```python
class Books:
  def __init__ (self,title,author,total_pages):
    self.title = title
    self.author = author
    self.total_pages = total_pages

  def __str__(self):
    return f" \"{self.title}\" by {self.author}"

book1 = Books("The Hobbit", "J. R. R. Tokien", 310)
book2 = Books("Harry Potter and The Philosopher's Stone", "J. K. Rowlling", 223)
book3 = Books("The Lion, the witch and the wardeobe", "C. S. Lewis", 172)

print(book2)
```

```python
from colorama import Fore, Back, Style, init

# Initialize colorama (autoreset automatically clears color after each print)
init(autoreset=True)

print(Fore.RED + 'This is red text!')
print(Fore.GREEN + Back.YELLOW + 'Green text with a yellow background!')
print(Style.BRIGHT + Fore.CYAN + 'Bright cyan text!')
```

```python
# Color codes
RED = "\033[31m"
GREEN = "\033[32m"
YELLOW = "\033[33m"
BLUE = "\033[34m"
RESET = "\033[0m"  # Resets formatting back to terminal default

print(f"{RED}This text is red!{RESET}")
print(f"{GREEN}This text is green!{RESET}")
```

```python
from termcolor import colored

print(colored('This is red text!', 'red'))
print(colored('Green text with a yellow background!', 'green', 'on_yellow'))
print(colored('Bright cyan text!', 'cyan', attrs=['bold']))
```

```python
from IPython.display import HTML, display

def print_color(text, color="red", bg_color="transparent"):
    display(HTML(f'<span style="color: {color}; background-color: {bg_color}; font-weight: bold;">{text}</span>'))

print_color("This is red text!", color="red")
print_color("Green text with a yellow background!", color="green", bg_color="yellow")
print_color("Bright cyan text!", color="cyan")
```
