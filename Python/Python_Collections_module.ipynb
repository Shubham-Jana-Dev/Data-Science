```python
from collections import namedtuple
student = namedtuple('Student',['name','age','marks'])
s1 = student('Shubham',20,85)
print(s1.name)
```

```python
print(s1.age)
```

```python
print(s1.marks)
```

```python
print(s1._fields)
```

```python
print(s1._asdict())
```

```python
print(s1._replace())
```

```python
print(s1._make((1,2,3)))
```

```python
print(s1.count(('Shubham',20,85)))
```

```python
print(s1.index(20))
```
