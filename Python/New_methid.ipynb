```python
class Point:
  def __init__(self,x,y):
    self.x = x
    self.y = y
  def __repr__(self):
    return f"Point({self.x},{self.y})"
p1 = Point(12,54)
print(p1.__repr__())
```

```python
class Person:
  def __init__(self,age):
    self.age = age
  def __str__(self):
    return self.age
u1 = Person(20)
print(u1.__str__())
```

```python
class M_Person:
  def __init__(self,name,age):
    self.name = name
    self.age = age
  def __str__(self):
    return f"{self.name} ({self.age})"
p3 = M_Person("Shubham",20)
print(p3.__str__())
```
