```python


class Format:
  def __init__(self,file_type,Hard_copy) -> None:
    self.file_type = file_type
    self.Hard_copy = Hard_copy

class Book:
  def __init__(self,Bname,Aname) -> None:
    self.Bname = Bname
    self.Aname = Aname
  def getTitle(self):
    return self.Bname

class eBook(Book):
  def __init__(self,book_name,author_name,file_Type = True,Hard_copy_present = False) -> None:
    super().__init__(book_name,author_name)
    self.file_type = Format(file_Type,Hard_copy_present)

class Library:
  def __init__(self,Book_collection) -> None:
    self.Book_collection = Book_collection

  def findBook(self,book_name):
    for i in range(0,len(self.Book_collection)):
      if(self.Book_collection[i].Bname == book_name):
        print(f"The book {book_name} written by {self.Book_collection[i].Aname} has found in row number {i + 1}. Kindly go to row number {i + 1} and collect the book \n Thank you 📚")
        break
    else:
        display(f"It looks like the Book {book_name} is not our invetory Soory :(")

  def addBook(self,new_Book):
    self.Book_collection.append(new_Book)
    display(f"The Book {new_Book.Bname} has successfully Added 📚")

myBook = Book('Growth Ultimate','Shubham Jana')
SrisiBook = Book('Life with CPP','Srisi Jana')
CBook = Book('C programin 5th edition',"Sunita Kar")


OurLibrary = Library([myBook,SrisiBook,CBook])
OurLibrary.findBook('Growth Ultimate')
akBook = Book("Ek kahani aishi Bhi","Bhoot uncle")
OurLibrary.addBook(akBook)
OurLibrary.findBook("Ek kahani aishi Bhi")
```

Create a system for a Library.

A Book has a title and author.

An EBook is a Book, but it also has a file size.

A Library has a collection of Books and methods to add or find a book by title.
