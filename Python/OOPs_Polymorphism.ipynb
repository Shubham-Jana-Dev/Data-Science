```python
# Function Overloading
class My_class:
  def display_info(self, subject = " " ):
    print(f"Hello {subject}")
new_year = My_class()
new_year.display_info(2026)
gretting = My_class()
gretting.display_info("Shubham")
```

```python
class parent:
  def displayinfo(self):
    print("Hi this is parent class 👨‍🦰")
class child(parent):
  def displayinfo(self):
    print("Hello this is child class 👶")
my_object = child()
my_object.displayinfo()
```

```python
class parent:
  def displayinfo(self):
    print("Hi this is parent class 👨‍🦰")
class child(parent):
  def displayinfo(self):
    super().displayinfo()
    print("Hello this is child class 👶")
child_object = child()
child_object.displayinfo()
```
