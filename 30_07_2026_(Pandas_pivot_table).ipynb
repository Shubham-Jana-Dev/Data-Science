```python
my_data = {
    "Employee Name":["Shubham Jana","Sujoy Roy","Rohan Bose","Dipak Jana","Pritam Ghorai","Nirmal Bera"],
    "Salary" : [45000,46000,70000,56000,60000,80000],
    "Gender" : ["Male","Male","Female","Male","Female","Male"],
    "Address": ["Kolkata","Mumbai","Delhi","Pune","Washington DC","Hydrabad"],
    "Department": ["IT","HR","Salse","IT","HR","Salse"],
    "Country" : ["India","Bangla Desh","India","USA","India","India"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(my_data)
```

```python
df
```
