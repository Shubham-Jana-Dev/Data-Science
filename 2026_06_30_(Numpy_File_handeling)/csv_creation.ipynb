```python
import numpy as np
roll = np.random.randint(34,56,20)
np.savetxt(
    'roll.txt',
    roll,
    fmt = '%d'
)
```

```python
np.savetxt(
    'roll_str.txt',
    roll,
    fmt = '%s'
)
```

```python
str_roll = np.loadtxt(
    'roll_str.txt',
    dtype = str
)
```

```python
print(str_roll)
np.savetxt(
    'ssttrr_roll.txt',
    str_roll,
    fmt = '%s'
)
```

```python
print(str_roll[1:4])
print(int(str_roll[5]))
```
