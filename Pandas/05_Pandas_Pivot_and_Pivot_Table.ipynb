```python
# Pivot table allows us to reshap and transform data :)
```

```python
import pandas as pd
```

```python
weather_data = {
    'date': ['1/1/2017','1/2/2017','1/3/2017','1/4/2017','1/5/2017','1/6/2017'],
    'city': ['new york','chicago','new york','new Delhi','new york','chicago'],
    'temperature': [65,66,75,72,71,75],
    'humidity': [21,22,35,24,32,30]
}
```

```python
df = pd.DataFrame(weather_data)
```

```python
display(df)
```

```python
df.pivot(index='date',columns='city')
```

```python
df.pivot(index='date',columns='city',values='temperature')
```

```python
df.pivot(index='humidity',columns='city')
```

```python
df.pivot(index='humidity',columns='city',values='temperature')
```

Pivot table is used to summarize of aggregate pivot data inside data frame

```python
weather2_data = {
    'date': ['1/1/2017','1/2/2017','1/3/2017','1/4/2017','1/5/2017','1/6/2017'],
    'city': ['new york','chicago','new york','chicago','new york','chicago'],
    'temperature': [65,66,75,72,71,75],
    'humidity': [21,22,35,24,32,30]
}
```

```python
df = pd.DataFrame(weather2_data)
```

```python
df.pivot_table(index='city',columns='date')
```

```python
df.pivot_table(index='city',columns='date',aggfunc='sum')
```

```python
df.pivot_table(index='city',columns='date',aggfunc='prod')
```

```python
df.pivot_table(index='city',columns='date',aggfunc='count')
```

The default aggregate function is mean :)

```python
df.pivot_table(index='city',columns='date',aggfunc='mean')
```

```python
df.pivot_table(index='city',columns='date',margins=False)
```

```python
df.pivot_table(index="city",columns="date",margins=True)
```

```python
weather3_data = {
    'date':['5/1/2017','5/2/2017','5/3/2017','12/4/2017','12/5/2017','12/6/2017'],
    'city':['new york','new york','new york','new york','new york','new york'],
    'temperature':[65,66,75,72,71,75],
    'humidity':[21,22,35,24,32,30]
}
```

```python
df = pd.DataFrame(weather3_data)
```

```python
display(df)
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
display(df)
```

```python
df.pivot_table(index=pd.Grouper(freq='ME',key='date'),columns='city')
```
