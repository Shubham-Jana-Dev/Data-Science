```python
class Calculator:
  def multiply(self, a = 1, b = 1, *args):
    result = a * b
    for num in args:
      result *= num
    return result
cla = Calculator()
print(cla.multiply())
print(cla.multiply(4))
print(cla.multiply(2,3))
print(cla.multiply(2,3,4))
```

```python
class  Animal:
  def awaj(self):
    return "Bol na!"
class Sudip(Animal):
  def awaj(self):
    return "Kai Kai..."
class Soumadip(Animal):
  def awaj(self):
    return "Chorr de mai Garibo ka mu nehi lagta..."
class Mannan(Animal):
  def awaj(self):
    return "Main kaha se fas jata hu har time..."
class Shubham(Animal):
  def awaj(self):
    return "Who Alaghi tha..."
class Tanmoy(Animal):
  def awaj(self):
    return "Dada ek Bat puche?.... "
class Soumalyo(Animal):
  def awaj(self):
    return "Amake ekta Chanar gilipi Dao na :("
students = [Sudip(), Soumadip(),Mannan(),Shubham(),Tanmoy(), Soumalyo()]
for i in students:
  print(i.awaj())
```

```python
class Car:
  def __init__(self,brand,model):
    self.brand = brand
    self.model = model
  def move(self):
    return "Bhrum.. Bhrum.. :)"
class Boat:
  def __init__(self,brand,model):
    self.brand = brand
    self.model = model
  def move(self):
    return "Chalak~~ Chalak~~ :)"
class Plane:
  def __init__(self,brand,model):
    self.brand = brand
    self.model = model
  def move(self):
    return "Sai-- Sai-- :)"

my_car = Car("Toyota","SUV")
my_plane = Plane("Air Bus","Boing 747")
my_boat = Boat("Hundai","Black suttle")

v = [my_car,my_plane,my_boat]

for i in v:
  print(i.move())
```

```python
class Vehicle:
  def __init__(self,brand,model):
    self.brand = brand
    self.model = model
  def move(self):
    return "Chal Na ... :)"
class Car(Vehicle):
  def move(self):
    return "Bhrum... bhrum... :)"
class Boat(Vehicle):
  def move(self):
    return "Chalak~~ Chalak~~ :)"
class Plane(Vehicle):
  def move(self):
    return "Suuuu--- :)"
my_car = Car("Tata", "Nano")
my_boat = Boat("Hundai", "Black suttle")
my_plane = Plane("Air Bus", "Boing 747")

gg = [my_car, my_boat, my_plane]
for i in gg:
  print(i.move())
```

```python
def add(dtype, *args):
  if(dtype == 'int'):
    res = 0
  elif(dtype == 'str'):
    res = ''
  for i in args:
    res += i
  print(res)
add('int',5,6)
add('str',"Hi ","Shubham")
```

```python
class Point:
  def __init__(self,x=0,y=0):
    self.x = x
    self.y = y
  def __str__(self):
    return f"({self.x},{self.y})"
  def __add__(self,other):
    x = self.x + other.x
    y = self.y + other.y
    return Point(x,y)
p1 = Point(1,2)
p2 = Point(2,3)
print(p1 + p2)
```

```python
class A:
  def __init__ (self,value):
    self.value = value
  def __add__(self, other):
    return self.value + other.value
ob1 = A(1)
ob2 = A(2)

print(ob1 +ob2)
print(A.__add__(ob1,ob2))
print(ob1.__add__(ob2))
print((10).__add__(100))
a = 23
b = 45
print(a.__add__(b))
```

```python
class complex:
  def __init__(self,real,img):
    self.real = real
    self.img = img
  def __add__(self,other):
    return self.real + other.real, self.img + other.img
c1 = complex(1,2)
c2 = complex(2,3)
print(c1 + c2)
```
