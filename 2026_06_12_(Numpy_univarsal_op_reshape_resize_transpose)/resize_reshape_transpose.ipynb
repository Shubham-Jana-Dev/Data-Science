```python
# Universal operator
import numpy as np
print(np.add(2,3))
print(np.subtract(5,6))
print(np.multiply(2,9))
print(np.divide(8,2))
print(np.floor_divide(77.4,5.77))
a = np.array([1.9,2.54,6.1,4.5,5.9])
print(np.max(a))
print(np.min(a))
print(np.square(5))
print(np.sqrt(25))
print(np.abs(-9.7))
print(np.abs(7.9))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.var(a))
print(np.mod(3,5))
print(np.power(a,3))

b = np.floor(a)
print(b.dtype)

```
