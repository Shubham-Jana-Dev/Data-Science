```python
import pandas as pd
import numpy as np
data = {
    'Name': ['Alice', 'Bob', 'Charlie', None,'David'],
    'Age': [25, None, 30, None,22],
    'Score': [85.5, 90.0, None, None, 88.0],
    'City': ['New York', 'Los Angeles', None, None,'Chicago']
}
```

```python
df = pd.DataFrame(data)
```

```python
display(df)
```

```python
dfe = df.dropna(how="any")
display(dfe)
```

```python
df["Score"] = df["Score"].fillna(0)
```
