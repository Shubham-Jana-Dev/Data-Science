# Probability Distribution
* np.random.normal(mean-value,scale,size)

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

arr1d = np.random.randint(10,50,10)
print(arr1d)
print(np.round(np.random.normal(50,10,10)))
```

* np.random.uniform(low,high,size)

```python
uniform_distribution = np.random.uniform(low=10, high=50, size=100000)
plt.figure(figsize=(7, 4))
sns.histplot(uniform_distribution, kde=False, discrete=True, color='salmon')
plt.title('Uniform Distribution (low=10, high=50, size=100000)')
plt.xlabel('Value')
plt.ylabel('frequency')
plt.show()
```

```python
print(np.round(np.random.uniform(10,20,4)))
binomial_data = np.random.binomial(n=12, p=0.2, size=10000)
plt.figure(figsize=(7, 4))
sns.histplot(binomial_data, kde=False, discrete=True, color='skyblue')
plt.title('Binomial Distribution (n=12, p=0.2)')
plt.xlabel('Number of Successes')
plt.ylabel('Frequency')
plt.show()
```

* np.random.uniform(number-trail,probability-of-success,size) <br>
=> 0 < probability-of-success < 1

```python
print(np.random.binomial(12,0.2,7))
```

## Poisson
* np.random.poisson(low(avarage occurance rate),size)

```python
print(np.random.poisson(34,9).reshape(3,1,3))
poisson_distribution = np.random.poisson(lam=34, size=10000)

# plot
plt.figure(figsize=(7, 4))
sns.histplot(poisson_distribution, kde=True, color='salmon')
plt.title("Poisson Distribution ($lamda=34$)")
plt.xlabel("Number of events")
plt.ylabel("Frequency")
plt.show()
```

```python
print(np.round(np.random.exponential(5,10)))

exponential_data = np.random.exponential(scale=5, size=10000)

# plot
plt.figure(figsize=(7,4))
sns.histplot(exponential_data, kde=True, color='lightgreen')
plt.title("Exponential Distribution (scale=5)")
plt.xlabel('Value / Time Between Events')
plt.ylabel('Frequency')
plt.show()
```

```python
time_between_calls = np.random.default_rng().exponential(scale=4, size=3)
print(time_between_calls)
```
