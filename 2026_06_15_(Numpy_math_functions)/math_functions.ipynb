```python
# Aggregate function
import numpy as np
arr = np.arange(20)
print(arr)

# np.sum()
print(np.sum(arr))
# np.average()
print(np.average(arr))
# np.absolute()
print(np.absolute(arr))
flo = np.array([12.44,5.76,-33.67,-78.90])
print(np.absolute(flo))
arr2d = arr.reshape(5,4)
print(arr2d)
# axis-based operations
print(np.sum(arr2d, axis = 1)) # while axis = 1 it returns the sum of the all elements of a single row 

print(np.sum(arr2d, axis = 0)) # while axis = 0 it returns the sum of the all elements of a single column


```

### Rounding functions 

* np.round() 

```python
float_arr = np.array([12.1113,34.555,3.456,6.789,55.675,-9.084,-19.9,-3.77])
print(np.round(float_arr,1))
```

* np.floor() 

```python
print(np.floor(float_arr))
```

* np.ceil()

```python
print(np.ceil(float_arr))
```

### Logarithmic functions 
* np.log()
* np.log10()
* np.log2()
* np.log().round(n)

```python
larr = [1,3.4,7,10,100,100]
print(np.log(larr))
print(np.log10(larr))
print(np.log2(larr))
print(np.log2(larr).round(1))
print(np.log1p(larr).round(2))
```

### Trignometric 
* np.sin()
* np.cos()
* np.tan()
* np.radians()
* np.arcsin()
* np.arccos()
* np.arctan()
* np.degrees()
