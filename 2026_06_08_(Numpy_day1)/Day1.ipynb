```python
import numpy as np
arr = np.array([16,34,64,75])
print(arr)
```

```python
# 2D array:
arr2d = np.array([['add','b','c'],['d','ettr','f']])
print(arr2d)
```

```python
# 3D array:
arr3d = np.array([[[1,2,3,4],[2,3,54,67],[11,54,76,89]],
                  [[12,44,65,78],[65,67,87,33],[23,45,12,13]],
                  [[30,54,21,43],[543,67,80,81],[122,554,789,10]]])
print(arr3d)
```

```python
# One's array:
print(np.ones(5))
# 2d ones array:
print(np.ones((2,3)))
# 3d ones array:
print(np.ones((3,3)))
```

```python
# Zero's array:
print(np.zeros(5))
# 2d zeros array:
print(np.zeros((2,3)))
# 3d zeros array:
print(np.zeros((3,3)))
```

```python
# Empty array:
print(np.empty(3))
# 2d empty array:
emp2d = np.empty((2,3))
print(emp2d)
# 3d empty array:
emp3d = np.empty((3,4))
print(emp3d)
```

```python
print(arr.shape)
print(arr2d.shape)
print(arr3d.shape)
```

```python
print(arr.size)
print(arr2d.size)
print(arr3d.size)
```

```python
print(arr.ndim)
print(arr2d.ndim)
print(arr3d.ndim)
```

```python
print(arr.dtype)
print(arr2d.dtype)
rr = np.array([[2.3,4.5,5.6],[2.4,3.5,7.9]])
print(rr.dtype)
```

```python
print(arr.itemsize)
print(arr2d.itemsize)
print(rr.itemsize)
```

```python
print(arr.nbytes)
print(arr2d.nbytes)
print(rr.nbytes)
```
