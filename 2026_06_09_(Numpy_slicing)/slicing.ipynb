```python
import numpy as np
a = np.array([10,20,40])
print(a.dtype)
```

```python
b = np.array([10.23,2.34,4.45])
print(b.dtype)
```

```python
c = np.array(['Shubham','Raj'])
print(c.dtype)
print(c.size)
print(c.itemsize)
```

```python
x = np.array(["Hello"],dtype = "U3")
print(x)
```

```python
y = np.array([-128],dtype = np.int8)
print(y)
```

```python
fl = np.array([12.33,34.5,65.67],dtype = np.int16)
print(fl)
```

```python
ff = a.astype(float)
print(ff)
print(a)
```
