```python
import numpy as np
arr1 = np.random.randint(1,11,4).reshape(2,2)

arr2 = np.random.randint(1,11,4).reshape(2,2)
print(arr1,"\n",arr2)
print(np.dot(arr1,arr2))
```

```python
print(np.matmul(arr1,arr2))
```
