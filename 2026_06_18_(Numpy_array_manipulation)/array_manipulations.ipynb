```python
import numpy as np
arr1d = np.arange(10)
print(arr1d)
arr1d2 = np.array([21,33,4,55,6,77,8,78,90,21])
arr2d = arr1d.reshape(2,5)
print(arr2d)

```

# Array manipulation
#### Array joining
----
* np.concatenate((a,b),axis = 0)

```python
catarr = np.concat((arr1d,arr1d2),axis = 0)
print(catarr)
```

```python
arr2d2 = np.array([[22,56,78,33,67],[12,13,15,67,87]])
print(arr2d,"\n",arr2d2)
p = np.concatenate((arr2d,arr2d2),axis = 0)
p1 = np.concatenate((arr2d,arr2d2),axis = 1)
print(p1,"\n\n",p)
```

* np.stack(())

```python
print(np.stack((arr1d,arr1d2)))
```

```python
print(np.hstack((arr1d,arr1d2)))
print(np.vstack((arr1d,arr1d2)))
```

* 2D implemention in np.hstack(())

```python
print(np.hstack((arr2d,arr2d2)))
print(np.vstack((arr2d,arr2d2)))
```

#### Spliting Array
* np.split(array_name,amount)

```python
print(np.split(arr1d,5))
print(np.split(arr1d,2))
```

* split() function in 2D array

```python
print(np.split(arr2d,2))
```

```python
print(np.array_split(arr1d,7))
print(np.array_split(arr1d2,6))
# --
print(np.array_split(arr2d[0],2))
```

```python
print(np.array_split(arr2d,2))
```

# Inserting Element
* np.insert()


```python
print(np.insert(arr1d,4,80))
```

```python
print(np.insert(arr2d2,[1],99))
print(arr2d2)
```

## Multiple values inserting
* np.append()

```python
print(np.append(arr1d,(233,555,678)))
print(np.append(arr2d2,(333,788,900)))
```

## Deleting element
* np.delete()

```python
print(np.delete(arr1d,0))
```

```python
print(np.delete(arr2d,2,1))
```
