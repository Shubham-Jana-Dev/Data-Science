## Manual indexing or Labeling:-

```python
subjects = pd.Series(["Mathematics","DBMS","DSA","DAA","COA"],index=[90,98,95,97,87])
subjects[98]
```

* .index

```python
subjects.index
```

* .values

```python
print(subjects.values)
```

* .dtypes

```python
print(subjects.dtype)
```

#### .size <br>
=> Returns the size of the Series..

```python
print(subjects.size)
```

#### .shape <br>
=> Returns the shape of an Series


```python
print(subjects.shape)
```

#### .loc[loc_value] <br>
=> Use to print a specific data when the user define index value (label value) is known

```python
print(subjects.loc[98])
```

#### .iloc[default_index_value]
=> Use to print a specific data when the default index value is known
