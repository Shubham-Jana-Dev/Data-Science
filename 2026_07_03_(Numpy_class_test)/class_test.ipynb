```python
import numpy as np

arr = np.random.randint(1, 100, (6, 6))
print("Original array:")
print(arr)
max_indices = np.argmax(arr, axis=1)
arr[np.arange(arr.shape[0]), max_indices] = 0

print("\nArray after replacing maximum elements with 0:\n")
print(arr)
```

```python
min_indices = np.argmin(arr, axis=0)
arr[np.arange(arr.shape[0]), min_indices] = -1
print(arr)
```

```python
# index = np.argwhere(arr % 2 == 0). # My code
even_rows = np.any(arr % 2 == 0, axis=1)
arr[even_rows] = np.sort(arr[even_rows], axis=1)
print(arr)
```

```python
# Teacher's code
arr[0::2] = np.sort(arr[0::2],axis = 1)
print(arr)
```

```python
np.save(
    "arr.npy",
    arr
)
x = np.load("arr.npy")
print(x)
```
