## 😷 Masked Array:-
* Masked array allows you to hide or ignore cirten values while performing calcualtion

```python
import numpy as np
import numpy.ma as ma
arr = np.array([(101,"xyz",80.9),(102,"pqr",72.6),(103,"ABC",85.0)],dtype=[("RollNo","i4"),("name","U10"),("marks","f4")])
print(arr["name"][1])
print(arr)
```

```python
data = np.array([3,-4,6,7,8,9])
masked = np.ma.masked_equal(data,-4)
print(masked)
```

```python
masked = np.ma.masked_where((data == -4)|(data == -7),data)
print(masked)
```
