```python
# Single Inheritance 📚
class Human:
  def can_eat():
    print("Human can eat :)")
  def can_run():
    print("Human can run :)")
class Our_teacher (Human):
  def __init__(self,name):
    self.name = name
  def can_teach(self):
    print(f"Our {self.name} sir can teach us. :)")
teacher = Our_teacher("Shubham")
Our_teacher.can_eat()
```

```python
# Multi level Inheritance 🤝
class Grand_Parent:
  def they_have(self):
    return "A Trackter"
class My_parent (Grand_Parent):
  def parent_has(self):
    return "A Computer"
class Me (My_parent):
  def I_have(self):
    return "A MacBook"
me = Me()
my_proparty = [me.I_have(),me.parent_has(),me.they_have()]
print(f"All I have are {my_proparty}.")
```

```python
# hierarchy Inheritance
class Animal:
  def __init__(self,name) -> None:
    self.name = name
  def can_breathe(self):
    print(f"{self.name} Can breathe")
  def can_run(self):
    print(f"{self.name} Can run")
class human(Animal):
  def __init__(self,name) -> None:
    super().__init__(name)
  def can_Code(self):
    print(f"{self.name} Can code and think logicaly.")
class Dog(Animal):
  def __init__(self,name) -> None:
    super(). __init__(name)
  def can_bark(self):
    print(f"{self.name} Can bark and always be hungry.")
Raj = human("Shubham")
print(f"what {Raj.name} can do are:-")
Raj.can_breathe()
Raj.can_run()
Raj.can_Code()
print()
Srisi = Dog("Srisi")
print(f"What {Srisi.name} can do, are:-")
Srisi.can_breathe()
Srisi.can_run()
Srisi.can_bark()

```

```python
class School:
  def __init__ (self,s_name) -> None:
    self.s_name = "JNNV"
class Student(School):
  def __init__ (self,name,s_name) -> None:
    self.name = name
    super().__init__(s_name)
st1 = Student("Shubham",'')
print(f"Name: {st1.name}\nSchool Name: {st1.s_name}")
```
