```python
# Single Inheritance 📚
class Human:
  def can_eat():
    print("Human can eat :)")
  def can_run():
    print("Human can run :)")
class Our_teacher (Human):
  def __init__(self,name):
    self.name = name
  def can_teach(self):
    print(f"Our {self.name} sir can teach us. :)")
teacher = Our_teacher("Shubham")
Our_teacher.can_eat()
```

```python
# Multi level Inheritance 🤝
class Grand_Parent:
  def they_have(self):
    return "A Trackter"
class My_parent (Grand_Parent):
  def parent_has(self):
    return "A Computer"
class Me (My_parent):
  def I_have(self):
    return "A MacBook"
me = Me()
my_proparty = [me.I_have(),me.parent_has(),me.they_have()]
print(f"All I have are {my_proparty}.")
```

```python
# hierarchy Inheritance
class Animal:
  def __init__(self,name) -> None:
    self.name = name
  def can_breathe(self):
    print(f"{self.name} Can breathe")
  def can_run(self):
    print(f"{self.name} Can run")
class human(Animal):
  def __init__(self,name) -> None:
    super().__init__(name)
  def can_Code(self):
    print(f"{self.name} Can code and think logicaly.")
class Dog(Animal):
  def __init__(self,name) -> None:
    super(). __init__(name)
  def can_bark(self):
    print(f"{self.name} Can bark and always be hungry.")
Raj = human("Shubham")
print(f"what {Raj.name} can do are:-")
Raj.can_breathe()
Raj.can_run()
Raj.can_Code()
print()
Srisi = Dog("Srisi")
print(f"What {Srisi.name} can do, are:-")
Srisi.can_breathe()
Srisi.can_run()
Srisi.can_bark()

```

```python
class School:
  def __init__ (self,s_name) -> None:
    self.s_name = "JNNV"
class Student(School):
  def __init__ (self,name,s_name) -> None:
    self.name = name
    super().__init__(s_name)
st1 = Student("Shubham",'')
print(f"Name: {st1.name}\nSchool Name: {st1.s_name}")
```

​The Problem: Smart Home & Room System
​The 'Is-A' Relationship: Create a base class called Device (with things like brand and status), and inherit it to make specialized classes like AC and SmartTV. That's your inheritance (is-a).
​The 'Has-A' Relationship: Create a Room class. A room has-a collection of devices inside it (like an AC object and a SmartTV object passed into it).

```python
class Device:
  def __init__ (self,brand,status) -> None:
    self.brand = brand
    self.status = status
class AC(Device):
  def __init__ (self,brand,status,name) -> None:
    self.name = name
    super(). __init__(brand,status)
class SmartTV(Device):
  def __init__ (self,brand,status,screen_type) -> None:
    self.screen_type = screen_type
    super(). __init__(brand,status)

class Room:
  def __init__ (self,woner_name,ac_b,ac_s,ac_n,tv_b,tv_s,tv_t):
    self.woner_name = woner_name
    self.ac_b = ac_b
    self.ac_s = ac_s
    self.tv_b = tv_b
    self.tv_s = tv_s
    self.tv_t = tv_t
    self.ac_n = ac_n
  def room_details(self) -> None:
    air_c = AC(self.ac_b,self.ac_s,self.ac_n)
    tv = SmartTV(self.tv_b,self.tv_s,self.tv_t)
    print(f"""
+----------------------------------+
| Room Woner: {self.woner_name}   \t   |
+----------------------------------+
|     AC Details:                  |
|        Brand:       {self.ac_b} \t   |
|        Status:      {self.ac_s} \t   |
|        AC Name:     {self.ac_n} \t   |
+----------------------------------+
|      TV Details:                 |
|        Brand:       {self.tv_b} \t   |
|        Status:      {self.tv_s} \t   |
|        Screen Type: {self.tv_t} \t   |
+----------------------------------+
    """)
won = "Shubham "
acb = "TATA"
acs = "ON"
acn = "Voltas"
tvb = "Samsung"
tvs = "OFF"
tvt = "LED"

our_room = Room(won,acb,acs,acn,tvb,tvs,tvt)
our_room.room_details()



```
