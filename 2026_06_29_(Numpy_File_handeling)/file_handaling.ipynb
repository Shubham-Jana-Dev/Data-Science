## File Handling:
* It's used when user wants to store arrays permanently and use them later without recreating them 

### save():
* It saves a single numpy array to a binary file with a .npy extention.

### savez():
* It saves multiple arrays to a single compressed file wiht a .npz extention. It's perfect when we have related arrays those are belong togater. (putting multiple related file into a zipfolder).

```python
import numpy as np
name = np.array(['x','y','z','a'])
age = np.array([20,15,16,15,21])
np.save("age.npy",age) 
```

```python
import numpy as np
class file_hadeling:
    def my_file_practice(self):
        name = np.array(['x','y','z','a'])
        age = np.array([20,15,16,15,21])
        np.save("age.npy",age)    
fh1 = file_hadeling()
fh1.my_file_practice()
    
```

```python
marks = np.array([56,78,90,42])
np.savez(
    'student.npz',
    age = age,
    name = name,
    marks = marks
)
print("Wo file ban geya")
```

## Load:
* Load is used to read arrays those were previusly saved in same our file.

```python
x = np.load("name.npy")
print(np.load('student.npz')['age'])
# print(vv['age'])
print(x)
```

```python
math = np.random.randint(67,89,10)
physics = np.random.randint(90,99,10)
bio = np.random.randint(45,78,10)
print(math,'\n',physics,'\n',bio)
np.savez(
    'Sujects.npz',
    math = math,
    physics = physics,
    bio = bio
)
print("File created successfully.")
```

```python
print(np.load('Sujects.npz')['math'] + np.load('Sujects.npz')['bio'])
print(np.median(np.load('Sujects.npz')['physics']))
```
