## Copy
* it's any dublicate data. It's means creating a new array with its won independent data. The original array and the copy array are completely separate. If we want to change in one array it doesn't effect the other one.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
arr1 = np.arange(1,11)
arr1c = arr1.copy()
print(arr1c)
print(id(arr1),"\n",id(arr1c))
plt.figure(figsize=(7, 4))
sns.histplot(arr1, kde=False, discrete=True, color='salmon')
plt.xlabel('Value')
plt.ylabel('frequency')
plt.show()
```

## View
* A view is a new array object that shear the same underline data as the original array, when we create a view we are not dublicating the data in memory instead we are creating a diffrent window or reference to the same data. 

```python
arr1v = arr1[0:5]
arr1[2] = 40
print(arr1)
print(arr1v)
```

## Shallow copy
* Shallow is when we create a new object, but it still referance the same underline data.
* in shallow copy only the array structure is copied not the actuall array data.

```python
arr1sh = arr1[:]
print(arr1 is arr1sh)
print(arr1sh.base is arr1)
arr1sh[0] = 8
print(arr1)
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(7, 4))
sns.lineplot(x=range(len(arr1)),y=arr1, marker="o",color="skyblue")
plt.xlabel('Position')
plt.ylabel('value')
plt.title("curve of Increasing Values")
plt.show()
```

```python
plt.figure(figsize=(7, 4))

# This automatically creates a cumulative growth curve of your data
sns.ecdfplot(data=arr1, color="salmon", marker="o")

plt.xlabel('Value')
plt.ylabel('Proportion of Data')
plt.title('Cumulative Increasing Curve')
plt.show()
```
