# Data cleaning:-

### Missing Values:-
* in pandas miossing data occurs when some values are missing or not collected properly this misssing values are represented as NaN, None.
* Pansa provides 2 importtant functions which help detecting missing values:-
* .isnull()
* returns a data frame of boolen values where true represent missing data

```python
import pandas as pd
data = {
    'Name': ['Amit','Priya', 'Rohit', None, 'Daa'],
    'Marks': [34,None,90,75,85]
}
df = pd.DataFrame(data)
print("Where value is persent\n")
df.isnull()
# df
```

#### df.isnull().sum() 
* count the missing values per column -> df.isnull().sum()
* count the total missing value in the data Frame -> df.isnull().sum()

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

##### .fillna(parameter)
* This function allow us to replace missing values with a specified value.
##### fillna()
* This is used replace missing values with a specific value

```python
df.fillna("Unknown")
```

```python
df["Marks"] = df["Marks"].fillna(56)
```

ffill(), bifill
