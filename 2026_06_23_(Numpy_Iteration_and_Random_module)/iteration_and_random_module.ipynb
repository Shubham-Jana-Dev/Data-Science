# Iteration
* for x in np.nditer(arr):


```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

arr1d = np.arange(1,11)
print(arr1d)
plt.figure(figsize=(7,4))
sns.histplot(arr1d, kde=False, discrete=True, color='salmon')
plt.title('1D array')
plt.show()
# iteration
for i in np.nditer(arr1d):
    print(i)

arr2d = np.arange(21,31).reshape(2,5)
print(arr2d)

# iteration in 2D array
for i in np.nditer(arr2d):
    print(i * 2)
```

# Read and Write
* for i in np.nditer(arr, op-flags = ['readwrite'])

```python
for x in np.nditer(arr1d, op_flags = ['readwrite']):
    x[...] = x *2
    print(x)
print(arr1d)


# read-write in 2D array
for s in np.nditer(arr2d, op_flags = ['readwrite']):
    s[...] = s * 1/2
print(arr2d)
```

# Iteration with index
* for index,value in np.ndenumerate(a):
    print(index,value)

```python
print("\n-v^v-v^v-In 1D Array-v^v-v^v-\n")
for index,value in np.ndenumerate(arr1d):
    print("index = ",index," value = ",value)

print("\n-v^v-v^v-In 2D Array-v^v-v^v-\n")
print(arr2d)
for i,v in np.ndenumerate(arr2d):
    print("index = ",i," value = ",v)

indices = np.arange(len(arr1d))

plt.scatter(indices, arr1d, color='crimson', s=100, zorder=3)

plt.title('Scatter Plot of Array Values by Index')
plt.xlabel('Inedx')
plt.ylabel('Value')
plt.xticks(indices)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
print("\n-v^v-v^v-In 3D Array-v^v-v^v-\n")
arr3d = np.arange(20,40).reshape(2,2,5)
for i1,v1 in np.ndenumerate(arr3d):
    print("index = ",i1,"value = ",v1)
```
