## Complex literals:
* When any rational number placed with the imaginary constant, that number behaves like complex number.
* The default value = 0j
* The member of complex literals always be belongs to < class 'coplex' >. We can define it by type() 

```python
x = complex()
print(x)
```

```python
y = complex(2)
print(y)
```

```python
z = complex(2,3)
print(z)
```

```python
print(complex(2,2.3))
print(complex(46.6,2.e54))
print(complex(4.5,3))
```

```python
print(2.e54)
```

```python
x = complex("123")
x
```

```python
print(complex("2.3"))
```

```python
print(complex("2+3j"))
```

```python
a = complex("2+4j")
b = complex("3+6j")
print(a+b)
print(b-a)
```

```python
print(complex("12+4j","1+2j"))
```

```python
print(complex(9,"4"))
```

```python
print(complex(7,int("5")))
```
